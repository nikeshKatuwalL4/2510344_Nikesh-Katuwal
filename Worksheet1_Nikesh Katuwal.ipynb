Problem - 1: Array Creation:

```python
import numpy as np

arr1 = np.empty((2, 2))
print("1. Empty array:\n", arr1)


arr2 = np.ones((4, 2))
print("\n2. Ones array:\n", arr2)


arr3 = np.full((3, 3), 7)
print("\n3. Full array (fill=7):\n", arr3)


a = np.array([[1, 2], [3, 4]])
arr4 = np.zeros_like(a)
print("\n4. Zeros like array a:\n", arr4)


arr5 = np.ones_like(a)
print("\n5. Ones like array a:\n", arr5)


new_list = [1, 2, 3, 4]
arr6 = np.array(new_list)
print("\n6. List to numpy array:\n", arr6)
```

Problem - 2: Array Manipulation: Numerical Ranges and Array indexing:

```python

arr1 = np.arange(10, 50)
print("\n1. Array 10 to 49:\n", arr1)


arr2 = np.arange(9).reshape(3, 3)
print("\n2. 3x3 matrix 0 to 8:\n", arr2)


arr3 = np.eye(3)
print("\n3. Identity matrix:\n", arr3)


arr4 = np.random.random(30)
print("\n4. Random array:\n", arr4)
print("Mean:", arr4.mean())


arr5 = np.random.random((10, 10))
print("\n5. Min:", arr5.min(), " Max:", arr5.max())


arr6 = np.zeros(10)
arr6[4] = 1
print("\n6. Replace 5th element with 1:\n", arr6)


arr7 = np.array([1, 2, 0, 0, 4, 0])
print("\n7. Reversed array:\n", arr7[::-1])


arr8 = np.ones((5, 5))
arr8[1:-1, 1:-1] = 0
print("\n8. Border 1 inside 0:\n", arr8)

arr9 = np.zeros((8, 8))
arr9[1::2, ::2] = 1
arr9[::2, 1::2] = 1
print("\n9. Checkerboard:\n", arr9)
```

Problem - 3: Array Operations:

```python
x = np.array([[1, 2], [3, 5]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])

print("\n1. Add:\n", x + y)
print("\n2. Subtract:\n", x - y)
print("\n3. Multiply by 3:\n", x * 3)
print("\n4. Square:\n", x ** 2)

print("\n5. Dot(v, w):", np.dot(v, w))
print("Dot(x, v):", np.dot(x, v))
print("Dot(x, y):\n", np.dot(x, y))

print("\n6. Concatenate x and y (rows):\n", np.concatenate((x, y), axis=0))
print("Concatenate v & w (columns):\n", np.vstack((v, w)))

print("\n7. Concatenate x and v (ERROR expected):")
print("Cannot concatenate because shapes do not match (2x2 vs length 2).")
```

Problem - 4: Matrix Operations:

```python
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

A_inv = np.linalg.inv(A)
print("\n1. A * A^-1:\n", A @ A_inv)

AB = A @ B
BA = B @ A
print("\n2. AB:\n", AB)
print("BA:\n", BA)
print("AB != BA =", not np.allclose(AB, BA))

lhs = (A @ B).T
rhs = B.T @ A.T
print("\n3. (AB)^T:\n", lhs)
print("B^T A^T:\n", rhs)

```

Linear equation using Inverse Methods.

```python
A = np.array([
    [2, -3, 1],
    [1, -1, 2],
    [3, 1, -1]
])

B = np.array([-1, -3, 9])

X = np.linalg.solve(A, B)
print("\nSolution to AX = B:\n x =", X[0], " y =", X[1], " z =", X[2])
```
