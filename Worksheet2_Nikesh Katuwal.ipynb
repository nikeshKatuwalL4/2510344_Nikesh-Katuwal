```python
import pandas as pd


medical_df = pd.read_csv("/content/drive/MyDrive/medical_students_dataset.csv")
medical_df.head()
```

2)


```python
medical_df.info()

print("\nMissing values per column:")
print(medical_df.isnull().sum())
```

3)

```python
num_cols = medical_df.select_dtypes(include=['int64','float64']).columns
cat_cols = medical_df.select_dtypes(include=['object']).columns

print("Numerical columns:", num_cols)
print("Categorical columns:", cat_cols)
```

Imputation Logic
Numerical columns = fill using mean (or median)

Categorical columns = fill using mode (most frequent value)

Imputing Numerical columns

```python
for col in num_cols:
    if medical_df[col].isnull().sum() > 0:
        medical_df[col] = medical_df[col].fillna(medical_df[col].mean())
        print(f"Filled missing values in numerical column '{col}' using MEAN.")
```

Imputing categorical columns

```python
for col in cat_cols:
    if medical_df[col].isnull().sum() > 0:
        medical_df[col] = medical_df[col].fillna(medical_df[col].mode()[0])
        print(f"Filled missing values in categorical column '{col}' using MODE.")
```

Mean for numerical columns
Because mean keeps the distribution smooth and works well when data is not heavily skewed.

Mode for categorical columns
Because categorical values do not have mathematical meaning. The most frequent category is the safest replacement.

4)

```python
print("Total duplicate rows:", medical_df.duplicated().sum())

# Remove duplicates
medical_df = medical_df.drop_duplicates()

print("Duplicates removed. New dataset size:", medical_df.shape)
```

```python
import pandas as pd

titanic = pd.read_csv("/content/drive/MyDrive/Titanic-Dataset.csv")
titanic.head()
```

```python
subset = titanic[['Name', 'Pclass', 'Sex', 'Age', 'Fare', 'Survived']]

first_class = subset[subset['Pclass'] == 1]

mean_fare = first_class['Fare'].mean()
median_fare = first_class['Fare'].median()
max_fare = first_class['Fare'].max()
min_fare = first_class['Fare'].min()

print("Fare Statistics for First Class Passengers:")
print("Mean Fare:", mean_fare)
print("Median Fare:", median_fare)
print("Maximum Fare:", max_fare)
print("Minimum Fare:", min_fare)
```

Problem 2:

```python
print("Null values in Age:", first_class['Age'].isnull().sum())


first_class_clean = first_class.dropna(subset=['Age'])

print("Shape after dropping missing Age:", first_class_clean.shape)
```

Problem 3:

```python

embarked_dummies = pd.get_dummies(titanic['Embarked'], prefix='Embarked')


titanic_encoded = pd.concat([titanic, embarked_dummies], axis=1)


titanic_encoded = titanic_encoded.drop(columns=['Embarked'])


titanic_encoded.head()
```

Problem 4:

```python
import matplotlib.pyplot as plt

survival_by_gender = titanic.groupby('Sex')['Survived'].mean()
print(survival_by_gender)

survival_by_gender.plot(kind='bar')
plt.title("Survival Rate by Gender")
plt.ylabel("Mean Survival Rate")
plt.xlabel("Gender")
plt.show()
```

Problem 5:

```python

survival_port_gender = titanic.groupby(['Embarked', 'Sex'])['Survived'].mean().unstack()
print(survival_port_gender)


survival_port_gender.plot(kind='bar')
plt.title("Survival Rate by Gender and Port of Embarkation")
plt.ylabel("Mean Survival Rate")
plt.xlabel("Port")
plt.show()
```
