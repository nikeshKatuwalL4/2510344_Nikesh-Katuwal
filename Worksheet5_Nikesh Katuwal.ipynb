```python
def cost_function(X, Y, W):
    n = len(Y)
    Y_pred = np.dot(X, W)
    return (1 / (2 * n)) * np.sum((Y_pred - Y) ** 2)
```

To do- 6

```python
def gradient_descent(X, Y, W, alpha, iterations):
    cost_history = [0] * iterations
    m = len(Y)

    for i in range(iterations):
        #Prediction
        Y_pred = np.dot(X, W)

        #Loss
        loss = Y_pred - Y

        #Gradient
        dw = (1 / m) * np.dot(X.T, loss)

        #Update weights
        W = W - alpha * dw

        #Cost
        cost_history[i] = cost_function(X, Y, W)

    return W, cost_history
```

To do- 8

```python
def rmse(Y, Y_pred):
    return np.sqrt(np.mean((Y - Y_pred) ** 2))
```

To do- 9

```python
def r2(Y, Y_pred):
    mean_y = np.mean(Y)
    ss_tot = np.sum((Y - mean_y) ** 2)
    ss_res = np.sum((Y - Y_pred) ** 2)
    return 1 - (ss_res / ss_tot)
```

To do- 10

```python
def main():
    # Load data
    dataset = pd.read_csv("/content/drive/MyDrive/student.csv")

    # Features and target
    X = dataset[['Math', 'Reading']].values
    Y = dataset['Writing'].values

    # Train-test split
    X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y, test_size=0.2, random_state=42
    )

    # Initialize weights
    W = np.zeros(X_train.shape[1])
    alpha = 0.00001
    iterations = 1000

    # Train model
    W_optimal, cost_history = gradient_descent(
        X_train, Y_train, W, alpha, iterations
    )

    # Predictions
    Y_pred = np.dot(X_test, W_optimal)

    # Evaluation
    model_rmse = rmse(Y_test, Y_pred)
    model_r2 = r2(Y_test, Y_pred)

    print("Final Weights:", W_optimal)
    print("Cost History (First 10):", cost_history[:10])
    print("RMSE:", model_rmse)
    print("R2 Score:", model_r2)
if __name__ == "__main__":
    main()
```
